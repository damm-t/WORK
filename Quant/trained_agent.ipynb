# TinyBrain v1

core concept :
ŷ = w * x + b

```python
w = 0.00
b = 0.00 

def model(x) :
    return w * x + b
```

```python
data = [
    (1,3),
    (2,5),
    (3,7),
    (4,9),
    (5,11),
]
```

```python
def loss(y_true, y_pred):
    return ( y_true - y_pred) ** 2
```

```python
lr_rate = 0.01

for epoch in range ( 1000 ):
    total_loss = 0

    for x,y in data :
        y_pred = model(x)
        error = y_pred - y
        print ("y in training : ", y , "||",  "error in training :", error )

        dw = error * x
        db = error

        w -= lr_rate* dw
        b -= lr_rate* db 

        total_loss += loss(y_pred , y)

if epoch % 100 == 0:
        print(epoch, total_loss)
```
